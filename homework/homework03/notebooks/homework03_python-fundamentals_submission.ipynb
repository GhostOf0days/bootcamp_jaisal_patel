```python
import sys

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Import helpers from src/.
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

# Raw CSV.
RAW_CSV = ROOT / "data" / "raw" / "starter_data.csv"
# Processed outputs.
PROCESSED_DIR = ROOT / "data" / "processed"
```

## NumPy operations

```python
# Sample prices.
prices = np.array([10, 12, 15, 18, 22])
scaled = prices * 1.1
centered = prices - prices.mean()

print("Original:", prices)
print("Scaled by 1.1 times:", scaled)
print("Centered:", centered)
```

## Loop vs. vectorization

```python
big_array = np.arange(1_000_000)

# Loop vs vectorized * 2.
%timeit [x * 2 for x in big_array]
%timeit big_array * 2
```

## Load and inspect dataset

```python
# Load starter CSV.
df = pd.read_csv(RAW_CSV)
df["date"] = pd.to_datetime(df["date"])

display(df.head())
df.info()
df.select_dtypes(include="number").describe()
```

## Summary statistics and groupby

```python
# Mean value by category.
summary = df.groupby("category", as_index=False)["value"].mean()
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
summary_path = PROCESSED_DIR / "summary.csv"
summary.to_csv(summary_path, index=False)

print(f"Saved: {summary_path}")
summary
```

# Plot

```python
# Value histogram.
df["value"].hist(bins=8)
plt.title("Value distribution")
plt.xlabel("value")
plt.ylabel("count")

plot_path = PROCESSED_DIR / "histogram.png"
plt.savefig(plot_path, bbox_inches="tight")
plt.show()
print(f"Saved: {plot_path}")
```
